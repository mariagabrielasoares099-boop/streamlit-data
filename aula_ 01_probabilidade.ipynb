# A aula 1 foi sobre a probabilidade dos acontecimentos de compras por categoria e a probabilidade dos acontecimentos acontecerem repetidos ou não.

```python
# Ativação das bibliotecas pandas e numpy python e chamar todas as tabelas e contexto
import pandas as pd
import numpy as np

df = pd.read_csv('superstore_sales.csv')
print(df.shape)
print(df.head(3))
```

```python
#Cálculo de pedidos de Tecnológia
total_pedidos = len(df)
pedidos_tech = (df['Category'] == 'Technology').sum()
prob_tech = pedidos_tech / total_pedidos

print("Total de pedidos:",total_pedidos)
print("Pedidos de Tecnologia:",pedidos_tech)
print("Probabilidade:", round(prob_tech,4))
```

```python
#Formula do pandas para calcular todas as contagens de vendas de uma vez só
prob_categorias = df['Category'].value_counts(normalize=True)
print(prob_categorias)
```

```python
#Complemento de cálculo de algo que não aconteceu(chance de 1%)
#Sempre lembrar que tem que considerar a regra de 1-P para o cálculo 
nao_tech_regra = 1 - prob_tech
nao_tech_direto = (df['Category'] != 'Technology').mean()

print("Pela regra (1-P):", round(nao_tech_regra, 4))
print("Contando direto:", round(nao_tech_direto, 4))
```

```python
# Regra de adição: eventos mutuamente excludentes (para pedidos de Technology ou Furniture)
# Como um pedido tem só uma categoria, somamos: P(A ou B) = P(A) + P(B)

prob_furn = (df['Category'] == 'Furniture').mean()

ou_exclusivo_regra = prob_tech + prob_furn
ou_exclusivo_direto = df['Category'].isin(['Technology', 'Furniture']).mean()

print('Pela regra (soma simples):', round(ou_exclusivo_regra, 4))
print('Contando direto (.isin):', round(ou_exclusivo_direto, 4))
```

```python
# Regra de adição geral: eventos que podem acontecer juntos (Technology OU West)
# Subtraímos P(Technology E West) para não contar a sobreposição duas vezes
mask_tech = df["Category"] == "Technology"
mask_west = df["Region"] == "West"

prob_west = mask_west.mean()
prob_tech_e_west = (mask_tech & mask_west).mean()

ou_geral_regra = prob_tech + prob_west - prob_tech_e_west
ou_geral_direto = (mask_tech | mask_west).mean()

print("Pela regra (soma - interseção):", round(ou_geral_regra, 4))
print("Contando direto (operador |):", round(ou_geral_direto, 4))
```

```python
# Probabilidade condicional: P(Technology | West)
# 1) Filtrando apenas pedidos da região West
cond_filtro = df.loc[mask_west, "Category"].eq("Technology").mean()

# 2) Pela definição formal: P(Technology E West) / P(West)
cond_formula = prob_tech_e_west / prob_west

print("P(Tech | West) via filtro:", round(cond_filtro, 4))
print("P(Tech | West) via fórmula:", round(cond_formula, 4))
print("P(Tech) geral:", round(prob_tech, 4))
```

```python
# Cada linha mostra a proporção das categorias dentro daquela região (soma = 1)
tabela_cond = pd.crosstab(df["Region"], df["Category"], normalize="index")
print(tabela_cond.round(4))
```
